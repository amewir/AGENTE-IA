Este ipnyb sirve para ordenar los pdf que se encuentran actualmente en la pagina de gae, con fecha maxima de subida en abril 2026

```python
import shutil
from pathlib import Path
import os


ruta_origen = r'C:\Users\angel.hernandez\Desktop\PGWL\Paginas-Web'
ruta_destino = r'C:\Users\angel.hernandez\Documents\Progra\Agente_IA_GAE\data\Docs\decreto'
```

```python
if not os.path.exists(ruta_destino):
    os.makedirs(ruta_destino)
    print(f"Se ha creado la carpeta, siendo su ruta: {ruta_destino}")
movido = 0
errores = 0

for raiz, dirs, archivos in os.walk(ruta_origen):
    for nombre_file in archivos:
        if "guia" in nombre_file.lower() and nombre_file.lower().endswith(".pdf"):
            ruta_full_origen = os.path.join(raiz, nombre_file)
            ruta_full_destino = os.path.join(ruta_destino, nombre_file)
            try: 
                shutil.move(ruta_full_origen, ruta_full_destino)
                movido +=1
            except Exception as e:
                print(f"Error encontrado {e}")
                errores +=1

```
